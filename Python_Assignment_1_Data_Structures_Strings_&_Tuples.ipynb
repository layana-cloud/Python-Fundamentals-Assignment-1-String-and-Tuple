Tuples

Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):

```python
#a) Concatenate the two tuples and store it in “t_combine”
t1=(10,20,30)
t2=(40,50,60)
t_combine =(t1+t2)
print(t_combine)
```

```python
#b) Repeat the elements of “t_combine” 3 times
print(t_combine*3)
```

```python
#c) Access the 3rd element from “t_combine”
print(t_combine[2])
```

```python
#d) Access the first three elements from “t_combine”
print(t_combine[0:3])
```

```python
#e) Access the last three elements from “t_combine”
print(t_combine[-3:])
```

t1 = (10,20,30) and t2 = (40,50,60)

a) t_combine = (t1+t2), + operator used to joins two tuples
* Result: (10,20,30,40,50,60)

b) t_combine*3
* 3 repeats the tuple 3 times
* Result: (10,20,30,40,50,60,10,20,30,40,50,60,10,20,30,40,50,60)

c) t_combine[2], to access the third element
* Index starts from 0
* Result: 30

d) t_combine[0:3], first 3 elements
* 0 to 2 = (10,20,30)

e) t_combine[-3:], Last 3 elements
* -3 means last 3 positions= (40,50,60)
